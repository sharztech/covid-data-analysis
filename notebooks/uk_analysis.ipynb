Import all necessary libraries

Sets parameters for a nice looking graph - optional

```python
sns.set(style="whitegrid") # Makes it clean
plt.rcParams["figure.figsize"] = (12,6) # Sets the default figure size
```

Load the cleaned version of the data gathered from the script

```python
# Load cleaned CSV
uk_df = pd.read_csv("../data/clean/uk_covid_clean.csv", parse_dates=["date"])
uk_df.head(20)
```

An overview into the New Daily Cases in the UK during the Pandemic

```python
# Creates a graph to show the daily new cases
plt.figure()
sns.lineplot(data=uk_df, x="date", y="new_cases")
plt.title("Daily New COVID-19 Cases in the UK")
plt.xlabel("Date")
plt.ylabel("New Cases")
plt.show()
```

Overview of new deaths in the UK

```python
plt.figure()
sns.lineplot(data = uk_df, x= "date", y="new_deaths")
plt.title("Daily new deaths in the UK (covid)")
plt.xlabel("Date")
plt.ylabel("Number of deaths")
plt.show()
```

Overview into new deaths vs new cases

```python
plt.figure()
sns.lineplot(data=uk_df, x="date", y=uk_df["new_cases"]/1_000_000, label="Daily Cases (millions)")
sns.lineplot(data=uk_df, x="date", y=uk_df["new_deaths"]/1_000_000, label="Daily Deaths (millions)")
plt.title("Comparison of Daily Cases vs Daily Deaths in the UK")
plt.xlabel("Date")
plt.ylabel("Count (millions)")
plt.legend()
plt.show()
```

An overview into the total deaths 

```python
# Creates a graph to show the total deaths
plt.figure()
sns.lineplot(data=uk_df, x="date", y="total_deaths")
plt.title("Daily Total Deaths in the UK")
plt.xlabel("Date")
plt.ylabel("Total Deaths")
plt.show()
```

Cummalative total deaths and total cases

```python
# Total deaths and total cases in the UK

plt.figure()
sns.lineplot(data=uk_df, x="date", y= uk_df["total_deaths"]/ 1000000, label = "Total Deaths")
sns.lineplot(data=uk_df, x="date", y=uk_df["total_cases"]/ 1000000, label = "Total Cases")
plt.title("Cummalative Total Deaths and Total Cases in the UK")
plt.xlabel("Date")
plt.ylabel("Count (millions)")
plt.legend()
plt.show()
```

Total Amount of Vaccinations in the UK

```python
# Showcases the total number of vaccinations

plt.figure()
sns.lineplot(data = uk_df, y="total_vaccinations", x ="date" )
plt.title("Total number of vaccinations in the UK")
plt.xlabel("Date")
plt.ylabel("Count")
plt.show()
```

Comparison of fully vaccinated to the not fully vaccinated people:

Graph 1: Count of people in millions

Graph 2: Percentaage of Population

```python
plt.figure()
sns.lineplot(data= uk_df, x= "date", y = uk_df["people_fully_vaccinated"]/1000000, label = "Fully Vaccinated")
sns.lineplot(data= uk_df, x= "date", y = uk_df["people_vaccinated"]/1000000, label = "One vaccination")
plt.title("Comparison of fully vaccinated people vs People with One vaccine")
plt.xlabel("Date")
plt.ylabel("Count (millions)")
plt.legend()
plt.show()
```

```python
# Vaccination percentages
uk_df["vaccinated_pct"] = (uk_df["people_fully_vaccinated"] / uk_df["population"]) * 100
uk_df["one_dose_pct"] = (uk_df["people_vaccinated"] / uk_df["population"]) * 100

plt.figure(figsize=(12,6))
sns.lineplot(data=uk_df, x="date", y="one_dose_pct", label="At least 1 dose", color="orange")
sns.lineplot(data=uk_df, x="date", y="vaccinated_pct", label="Fully vaccinated", color="green")
plt.title("COVID-19 Vaccination Progress in the UK (% of population)")
plt.xlabel("Date")
plt.ylabel("Percentage of population (%)")
plt.legend()
plt.show()
```

Subplot graph of Daily cases and deaths and new vaccinations

```python
# Create a figure with 3 rows and 1 column of subplots
fig, axes = plt.subplots(3, 1, figsize=(14, 12), sharex=True)

# 1. Daily cases
sns.lineplot(data=uk_df, x="date", y="new_cases", ax=axes[0], color="blue")
axes[0].set_title("Daily COVID-19 Cases (UK)")
axes[0].set_ylabel("Cases (millions)")

# 2. Daily deaths
sns.lineplot(data=uk_df, x="date", y="new_deaths", ax=axes[1], color="red")
axes[1].set_title("Daily COVID-19 Deaths (UK)")
axes[1].set_ylabel("Deaths")

# 3. Vaccinations
sns.lineplot(data=uk_df, x="date", y=uk_df["people_fully_vaccinated"]/1_000_000, ax=axes[2], color="green")
axes[2].set_title("Fully Vaccinated People (UK)")
axes[2].set_ylabel("Millions")
axes[2].set_xlabel("Date")

# Make sure titles and labels don’t overlap
plt.tight_layout()
plt.show()
```

Merged Graph for Example:

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, ax1 = plt.subplots(figsize=(14,6))

# Plot daily cases on left y-axis
sns.lineplot(data=uk_df, x="date", y="new_cases", ax=ax1, label="Daily Cases", color="blue")
ax1.set_ylabel("Daily Cases")
ax1.set_xlabel("Date")

# Plot daily deaths on same left axis or separate right axis
ax2 = ax1.twinx()
sns.lineplot(data=uk_df, x="date", y="new_deaths", ax=ax2, label="Daily Deaths", color="red")
ax2.set_ylabel("Daily Deaths")

# Plot fully vaccinated people on secondary axis scaled (millions)
ax3 = ax1.twinx()
ax3.spines['right'].set_position(('outward', 60))  # Move third axis to avoid overlap
sns.lineplot(data=uk_df, x="date", y=uk_df["people_fully_vaccinated"]/1_000_000, ax=ax3, label="Fully Vaccinated (millions)", color="green")
ax3.set_ylabel("Fully Vaccinated (millions)")

ax2.legend_.remove()  # remove auto legend from ax2
ax3.legend_.remove()  # remove auto legend from ax3

# Combine legends
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
lines3, labels3 = ax3.get_legend_handles_labels()
ax1.legend(lines1 + lines2 + lines3, labels1 + labels2 + labels3, loc="upper left")

plt.title("COVID-19 UK: Daily Cases, Daily Deaths, and Vaccination Progress")
plt.show()
```
